```python
#Annotate orthologs against eggNOG database
mosh annotate map-eggnog \
    --i-eggnog-hits ./eggnog_hits.qza \
    --i-db ./eggnog_db.qza \
    --p-num-cpus 128 \
    --p-db-in-memory \
    --o-ortholog-annotations ./eggnog_annotations.qza \
    --verbose
```

```python
#Extract annotations
mosh annotate extract-annotations \
    --i-ortholog-annotations ./eggnog_annotations.qza \
    --p-annotation caz \
    --p-max-evalue 0.0001 \
    --o-annotation-frequency ./caz_annot_ft.qza \
    --verbose
```

```python
#Multiply tables
mosh annotate multiply-tables \
    --i-table1 ./mags_derep_all_domains.qza \
    --i-table2 ./caz_annot_ft.qza \
    --o-result-table ./caz_ft.qza \
    --verbose
```

```python
#calculating a Bray-curtis dissimilarity matrix to measure the dissimilarity between each sample, based on observed frequency of different CAZyme annotations in each sample.
mosh diversity beta \
    --i-table ./caz_ft.qza \
    --p-metric braycurtis \
    --o-distance-matrix ./caz_braycurtis_dist.qza
```

```python
#Principal coordinate analysis (PCoA)
mosh diversity pcoa \
    --i-distance-matrix ./caz_braycurtis_dist.qza  \
    --o-pcoa ./caz_braycurtis_pcoa.qza
```

```python
#plot the PCoa results
mosh emperor plot \
    --i-pcoa ./caz_braycurtis_dist.qza \
    --m-metadata-file ./updog_metadata.tsv \
    --o-visualization caz-pcoa.qzv
```
